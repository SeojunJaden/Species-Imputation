```python
import pandas as pd
import numpy as np
import requests
from global_land_mask import globe
pd.set_option('display.max_columns', None)
pd.set_option('display.max_columns', None)
mission_bay = pd.read_csv('C:/Users/maxph/OneDrive/Desktop/SpeciesImputation/MissionBayData.csv')
mission_bay = mission_bay.drop(['url', 'image_url', 'sound_url', 'tag_list', 'description', 'captive_cultivated', 'oauth_application_id', 'private_place_guess', 'private_latitude', 'private_longitude', 'public_positional_accuracy', 'geoprivacy', 'taxon_geoprivacy', 'coordinates_obscured', 'positioning_method', 'positioning_device', 'species_guess', 'created_at', 'updated_at'], axis = 1)
mission_bay
```

```python
mission_bay['time_observed_at'] = pd.to_datetime(mission_bay['time_observed_at'])
mission_bay = mission_bay.sort_values(by= ['scientific_name', 'time_observed_at'])
mission_bay
```

```python
def filter_observations(group):
    kept_rows = []
    last_kept_time = None
    
    for _, row in group.iterrows():
        # If it's the first time seeing this species OR it's been > 5 mins
        if last_kept_time is None or (row['time_observed_at'] - last_kept_time) >= pd.Timedelta(minutes=5):
            kept_rows.append(row)
            last_kept_time = row['time_observed_at']
            
    return pd.DataFrame(kept_rows)

mission_bay_clean = mission_bay.groupby('scientific_name', group_keys = False).apply(filter_observations)
mission_bay_clean
```

```python
import asyncio
import httpx

async def get_habitat_async(client, scientific_name):
    # Match Name
    match_url = f"https://api.gbif.org/v1/species/match?name={scientific_name}"
    resp = await client.get(match_url)
    usage_key = resp.json().get('usageKey')

    if not usage_key: return "Name not found"

    # Get Profile
    profile_url = f"https://api.gbif.org/v1/species/{usage_key}/speciesProfiles"
    profile_resp = await client.get(profile_url)
    results = profile_resp.json().get('results', [])

    if not results: return "No habitat data"
    
    info = results[0]
    habitats = [h for h in ['marine', 'freshwater', 'terrestrial'] if info.get(h)]
    return ", ".join(habitats).capitalize() if habitats else "Unknown"

async def main(names):
    async with httpx.AsyncClient() as client:
        tasks = [get_habitat_async(client, name) for name in names]
        return await asyncio.gather(*tasks)
```

```python
mission_bay_final = mission_bay_clean.assign(coordinates = np.array(zip(mission_bay_clean.get('latitude'), mission_bay_clean.get('longitude'))))
mission_bay_final['is_land'] = mission_bay_final['coordinates'].apply(lambda x: globe.is_land(x[0], x[1]))
mission_bay_final
```

```python
mission_bay_final['habitat'] = mission_bay_final['scientific_name'].apply(get_habitat)
```
